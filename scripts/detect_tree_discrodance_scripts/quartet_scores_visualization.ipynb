```python
import toytree
import toyplot
import toyplot.pdf
import pandas as pd
import toyplot.svg
```

```python
# read a tree
rtre = toytree.io.tree("../quartet_scores_output/reformatted_quartet_scores_output.nwk",internal_labels="LQ-IC/QP-IC/EQP-IC")
# rtre = tre.root("SRR11748982")
```

```python
# get dictionary where key is the accession number and value is the species name

# read the csv file
df = pd.read_csv('/Users/yue/Documents/research/0-data/IMAP.tsv', delimiter="\t", header=None)

# convert df to dictionary
tip_dict = pd.Series(df.iloc[:, 1].values, index=df.iloc[:, 0]).to_dict()
```

# Extract LQ_IC, QP_IC and EQP_IC

```python
# create three empty lists to store the corresponding values
LQ_IC = []
QP_IC = []
EQP_IC = []

# iterate each node to extract these three values
for i in rtre.get_node_data("LQ-IC/QP-IC/EQP-IC"):
    # if the string is like "0.122204/0.037934/0.091647/0.999976"
    # the first value is LQ-IC, the second value is QP-IC, the third value is EQP-IC and the last value is support
    if type(i) == str and len(i.split("/")) == 4:
        LQ_IC.append(round(float(i.split("/")[0]),1))
        QP_IC.append(round(float(i.split("/")[1]),1))
        EQP_IC.append(round(float(i.split("/")[2]),1))
    else:
        LQ_IC.append("nan")
        QP_IC.append("nan")
        EQP_IC.append("nan")

# map these three values as three distinct features
rtre= rtre.set_node_data(feature = "LQ-IC", data = LQ_IC)
rtre= rtre.set_node_data(feature = "QP-IC", data = QP_IC)
rtre= rtre.set_node_data(feature = "EQP-IC", data = EQP_IC)
```

```python
# use colors to distinguish internal nodes with high and low values
# dark green "#2F4F4F", for values >= 0.5
# green "#98F898", for values betwen 0.5 and 0
# pale red "#FFCC99" for values between -0.4 and 0
# red "#FF6600 for values < -0.4
EQP_IC_colors = [
    "#2F4F4F" if float(i) >= 0.5 else
    "#98F898" if 0 <= float(i) < 0.5 else
    "#FFCC99" if -0.4 <= float(i) < 0 else
    "#FF6600"
    for i in rtre.get_node_data('EQP-IC')
]
```

```python
rtre = toytree.mod.ladderize(rtre, direction = False)
```

```python
canvas, axes, makr = rtre.draw(
    tip_labels=(tip_dict[i] for i in rtre.get_tip_labels()),
    tip_labels_align=True,
    node_labels="EQP-IC",
    node_sizes=14,
    node_colors=EQP_IC_colors,
    tree_style='n',
    width = 1000
);
```

```python
toyplot.pdf.render(canvas, "../quartet_scores_output/tree_EQP_IC_values.pdf")
```

```python
toyplot.svg.render(canvas, "../quartet_scores_output/tree_EQP_IC_values.svg")
```
